```python
#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

#Import Excel file
df = pd.read_excel('applications.xlsx')
```

```python
#Matplotlib Task 1

#Count applicants by college
college_counts = df['College'].value_counts().sort_index()

#Create bar plot
plt.bar(college_counts.index, college_counts.values, color='skyblue')

#Add title and labels
plt.title('Number of Applications by College')
plt.xlabel('College')
plt.ylabel('Number of Applicants')

plt.show()
```

```python
#Seaborn Task 1

#Create frequency bar plot
sns.countplot(data=df, x='College', color='lightgreen')

#Add title and labels
plt.title('Number of Applications by College')
plt.xlabel('College')
plt.ylabel('Number of Applicants')

plt.show()
```

```python
#Plotly Task 1

#Count applicants by college
college_counts = df['College'].value_counts().sort_index()

#Create interactive bar plot
fig = px.bar(
    x=college_counts.index,
    y=college_counts.values,
    color_discrete_sequence=['orange'],
    title='Number of Applications by College',
    labels={'x': 'College',
            'y': 'Number of Applicants'}
)

fig.show()
```

```python
#matplotlib task 2

#Import libraries
import pandas as pd
import matplotlib.pyplot as plt

#Define class intervals
bins = [800, 900, 1000, 1100, 1200, 1300, 1400]

#Create histogram and get frequencies
plt.figure(figsize=(8, 5))
frequency, intervals, bars = plt.hist(
    df['SAT_ACT'], bins=bins, color='skyblue', edgecolor='black'
)

#Display frequency distribution
print("Class Intervals:", intervals)
print("Frequencies:", frequency)

#Add title and axis labels
plt.title('Distribution of SAT_ACT Scores')
plt.xlabel('SAT_ACT Score')
plt.ylabel('Frequency')

plt.show()
```

```python
#Seaborn Task 2

#Import libraries
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

#Define class intervals
bins = [800, 900, 1000, 1100, 1200, 1300, 1400]

#Calculate frequency distribution
frequency, intervals = np.histogram(df['SAT_ACT'].dropna(), bins=bins)

#Display frequency distribution
print("Class Intervals:", intervals)
print("Frequencies:", frequency)

#Create histogram using Seaborn
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='SAT_ACT', bins=bins,
             color='skyblue', edgecolor='black')

#Add title and axis labels
plt.title('Distribution of SAT_ACT Scores')
plt.xlabel('SAT_ACT Score')
plt.ylabel('Frequency')

plt.show()
```

```python
#Plotly Task 2

#Import libraries
import plotly.express as px
import numpy as np

#Define six class intervals
bins = [800, 900, 1000, 1100, 1200, 1300, 1400]

#Calculate frequency distribution
frequency, intervals = np.histogram(df['SAT_ACT'].dropna(), bins=bins)

#Display frequency distribution
print("Class Intervals:", intervals)
print("Frequencies:", frequency)

#Create histogram using Plotly
fig = px.histogram(
    df,
    x='SAT_ACT',
    nbins=6,
    color_discrete_sequence=['blue'],
    title='Distribution of SAT_ACT Scores',
    width=800,
    height=500
)

#Set interval width and starting point
fig.update_traces(xbins=dict(start=800, end=1400, size=100))

#Add axis labels
fig.update_layout(
    xaxis_title='SAT_ACT Score',
    yaxis_title='Frequency'
)

fig.show()
```

```python
#matplotlib task 3

#Import libraries
import matplotlib.pyplot as plt
import numpy as np

#Create a scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df['SAT_ACT'], df['Extracurricular_Hours'], color='blue')

#Add a linear trendline
m, b = np.polyfit(df['SAT_ACT'], df['Extracurricular_Hours'], 1)
plt.plot(df['SAT_ACT'], m * df['SAT_ACT'] + b, color='red')

#Add title and axis labels
plt.title('SAT_ACT Scores vs Extracurricular Hours')
plt.xlabel('SAT_ACT Score')
plt.ylabel('Extracurricular Hours')

plt.show()
```

```python
#seaborn task 3

#Import libraries
import seaborn as sns
import matplotlib.pyplot as plt

#Create a scatter plot
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='SAT_ACT', y='Extracurricular_Hours',
            scatter_kws={'color': 'blue'},
            line_kws={'color': 'red'},
            ci=None)

#Add title and axis labels
plt.title('SAT_ACT Scores vs Extracurricular Hours')
plt.xlabel('SAT_ACT Score')
plt.ylabel('Extracurricular Hours')

plt.show()
```

```python
#Plotly task 3

#Import Plotly
import plotly.express as px

#Create scatter plot
fig = px.scatter(
    df,
    x='SAT_ACT',
    y='Extracurricular_Hours',
    trendline='ols',
    color_discrete_sequence=['blue'],
    title='SAT_ACT Scores vs Extracurricular Hours',
    width=800,
    height=500
)

#Add axis labels
fig.update_layout(
    xaxis_title='SAT_ACT Score',
    yaxis_title='Extracurricular Hours'
)

fig.show()
```
