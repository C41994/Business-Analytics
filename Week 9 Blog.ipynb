```python
# sqlite3 provides tools for connecting to SQLite databases.
import sqlite3
# pandas works with tables; pd is its short name.
import pandas as pd

# Connect to the database (creates the file if it doesn't exist)
conn = sqlite3.connect('my_data.db')

# Create a cursor object to execute SQL commands
cursor = conn.cursor()
```

```python
pd.read_csv('sales.csv').to_sql('sales',conn,if_exists='replace',index=False)
```

```python
pd.read_sql_query('''
SELECT * FROM sales
''',conn)
```

```python
pd.read_sql_query('''
SELECT sale_date, product, price FROM sales
''',conn)
```

```python
pd.read_csv('restaurant_orders.csv').to_sql('Orders', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT order_date, ROUND(SUM(order_total), 2) AS total_orders
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

** Task 6

```python
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT doctor,
       ROUND(AVG(patient_age), 1) AS patient_age,
       MAX(wait_minutes) AS wait_minutes
FROM visits
GROUP BY doctor
''', conn)
```

** Task 7

```python
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query('''
SELECT c.branch,
SUM(t.amount) AS total_transaction_amount
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
